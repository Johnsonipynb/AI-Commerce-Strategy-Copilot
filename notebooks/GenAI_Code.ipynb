```python
import time

def safe_generate(prompt):

    while True:
        try:
            response = model.generate_content(prompt)
            return response.text

        except Exception as e:

            if "429" in str(e):
                print("Rate limit reached. Waiting 40 seconds...")
                time.sleep(40)

            else:
                raise e
```

```python
prompt = f"""
You are a Chief Ecommerce Strategy Officer.

Analyze the following ecommerce business metrics.

{business_summary}

Provide:

1. Executive Summary
2. Top Revenue Risks
3. Recommended Actions
4. Growth Opportunities

Use professional executive language.
"""

result = safe_generate(prompt)

print(result)
```
